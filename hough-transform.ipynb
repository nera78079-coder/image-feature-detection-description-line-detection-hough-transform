```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
border = cv2.imread('border.jpg')
```

```python
building = cv2.imread('building.jpg')
```

```python
def hough_transform (img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    lines = cv2.HoughLines(edges, 1, np.pi/180, 100)
    for line in lines:
        rho, theta = line[0]
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
    
        x1 = int(x0 + 1000*(-b))
        y1 = int(y0 + 1000*(a))
        x2 = int(x0 - 1000*(-b))
        y2 = int(y0 - 1000*(a))
    
        cv2.line(gray, (x1,y1), (x2,y2), (0,0,255), 2)
    
    plt.imshow(gray,cmap="gray")
```

```python
hough_transform(border)
```

```python
hough_transform(building)
```
